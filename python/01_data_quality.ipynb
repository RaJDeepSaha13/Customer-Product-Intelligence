```python
import pandas as pd
import numpy as np
import os

print("Python is working!")
print("Pandas version:", pd.__version__)
```

```python
data_path = "../data"

customers = pd.read_csv(f"{data_path}/customers.csv")
products = pd.read_csv(f"{data_path}/products.csv")
orders = pd.read_csv(f"{data_path}/orders.csv")
order_items = pd.read_csv(f"{data_path}/order_items.csv")
sessions = pd.read_csv(f"{data_path}/sessions.csv")
user_events = pd.read_csv(f"{data_path}/user_events.csv")
operations = pd.read_csv(f"{data_path}/operations.csv")
support_tickets = pd.read_csv(f"{data_path}/support_tickets.csv")
experiments = pd.read_csv(f"{data_path}/experiments.csv")

print("All 9 datasets loaded successfully!")
```

```python
tables = {
    "Customers": customers,
    "Products": products,
    "Orders": orders,
    "Order Items": order_items,
    "Sessions": sessions,
    "User Events": user_events,
    "Operations": operations,
    "Support Tickets": support_tickets,
    "Experiments": experiments
}

for name, df in tables.items():
    print(f"{name}: {df.shape[0]:,} rows × {df.shape[1]} columns")
```

```python
for name, df in tables.items():
    print("\n" + "=" * 50)
    print(name)
    print("=" * 50)
    print(df.isnull().sum())
```

```python
for name, df in tables.items():
    missing = df.isnull().sum()
    missing = missing[missing > 0]

    print("\n" + "=" * 50)
    print(name)
    print("=" * 50)

    if len(missing) == 0:
        print("No missing values")
    else:
        print(missing)
```

```python
for name, df in tables.items():
    duplicates = df.duplicated().sum()
    print(f"{name}: {duplicates:,} duplicate rows")
```

```python
for name, df in tables.items():
    print("\n" + "=" * 50)
    print(name)
    print("=" * 50)
    print(df.dtypes)
    
```

```python
# Convert date columns to datetime

customers["signup_date"] = pd.to_datetime(customers["signup_date"])
products["launch_date"] = pd.to_datetime(products["launch_date"])
orders["order_date"] = pd.to_datetime(orders["order_date"])
sessions["session_date"] = pd.to_datetime(sessions["session_date"])
user_events["event_date"] = pd.to_datetime(user_events["event_date"])
operations["delivery_date"] = pd.to_datetime(operations["delivery_date"])
operations["promised_date"] = pd.to_datetime(operations["promised_date"])
support_tickets["ticket_date"] = pd.to_datetime(support_tickets["ticket_date"])
experiments["assignment_date"] = pd.to_datetime(experiments["assignment_date"])

# Keep product IDs consistently as text
products["product_id"] = products["product_id"].astype("string")
order_items["product_id"] = order_items["product_id"].astype("string")
user_events["product_id"] = user_events["product_id"].astype("string")

print("Date and ID columns converted successfully.")
```

```python
# Location of our original raw data
data_path = "../data/raw"

customers = pd.read_csv(f"{data_path}/customers.csv")
products = pd.read_csv(f"{data_path}/products.csv")
orders = pd.read_csv(f"{data_path}/orders.csv")
order_items = pd.read_csv(f"{data_path}/order_items.csv")
sessions = pd.read_csv(f"{data_path}/sessions.csv")
user_events = pd.read_csv(f"{data_path}/user_events.csv")
operations = pd.read_csv(f"{data_path}/operations.csv")
support_tickets = pd.read_csv(f"{data_path}/support_tickets.csv")
experiments = pd.read_csv(f"{data_path}/experiments.csv")

print("Raw datasets loaded successfully!")
```

```python
# Create clean copies
customers_clean = customers.copy()
products_clean = products.copy()
orders_clean = orders.copy()
order_items_clean = order_items.copy()
sessions_clean = sessions.copy()
user_events_clean = user_events.copy()
operations_clean = operations.copy()
support_tickets_clean = support_tickets.copy()
experiments_clean = experiments.copy()

print("Clean copies created.")
```

```python
# Convert date columns to proper datetime format

customers_clean["signup_date"] = pd.to_datetime(
    customers_clean["signup_date"]
)

products_clean["launch_date"] = pd.to_datetime(
    products_clean["launch_date"]
)

orders_clean["order_date"] = pd.to_datetime(
    orders_clean["order_date"]
)

sessions_clean["session_date"] = pd.to_datetime(
    sessions_clean["session_date"]
)

user_events_clean["event_date"] = pd.to_datetime(
    user_events_clean["event_date"]
)

operations_clean["delivery_date"] = pd.to_datetime(
    operations_clean["delivery_date"]
)

operations_clean["promised_date"] = pd.to_datetime(
    operations_clean["promised_date"]
)

support_tickets_clean["ticket_date"] = pd.to_datetime(
    support_tickets_clean["ticket_date"]
)

experiments_clean["assignment_date"] = pd.to_datetime(
    experiments_clean["assignment_date"]
)

print("Date columns cleaned.")
```

```python
products_clean["product_id"] = products_clean["product_id"].astype("string")

order_items_clean["product_id"] = (
    order_items_clean["product_id"].astype("string")
)

user_events_clean["product_id"] = (
    user_events_clean["product_id"].astype("string")
)

print("Product IDs standardized.")
```

```python
import os

cleaned_path = "../data/cleaned"

os.makedirs(cleaned_path, exist_ok=True)
```

```python
clean_tables = {
    "customers": customers_clean,
    "products": products_clean,
    "orders": orders_clean,
    "order_items": order_items_clean,
    "sessions": sessions_clean,
    "user_events": user_events_clean,
    "operations": operations_clean,
    "support_tickets": support_tickets_clean,
    "experiments": experiments_clean
}

for name, df in clean_tables.items():
    df.to_csv(
        f"{cleaned_path}/{name}.csv",
        index=False
    )

print("All cleaned datasets saved successfully!")
```
